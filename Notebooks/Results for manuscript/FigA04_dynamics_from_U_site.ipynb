# Fig. A4 — Stroboscopic dynamics in a long chain, particle starting on a U site

Stroboscopic occupations $P_{\alpha_i}(mT) = |\langle \alpha_i | \hat U_{\rm t}(mT) | \psi_0 \rangle|^2$ for a closed chain with
$N = 100$ cells ($301$ sites), antisymmetric driving ($\zeta = -1$), $\omega = 8$, $\tilde A = 1.2$ ($A = 9.6$), $J = 1$, $\phi = \pi$,
over 400 driving cycles.
Computed with the exact Floquet operator of the square wave (`stroboscopic_populations_exact`).

Initial state: the central $U$ site.

> Labelling: with sites $C_1, \dots, C_{101}$ the central site is $U_{51}$ (0-based site index 151). The figure labels
> cells starting from 1, as Figs. 4, 5, 6 and A1.

Output: `Data/FigA04_dynamics_U_site.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from diamond_chain import site_index, stroboscopic_populations_exact
from data_io import save_data
```

```python
J = 1.0
phi = np.pi
Ncells = 100
w = 8
A = 9.6
zeta = -1
ncycles = 400

initial_site = site_index('U', 51)      # U_51, central cell of the chain
print('initial site index:', initial_site)
```

```python
# Exact Floquet operator of the square wave (two matrix exponentials per period)
tvals, populations = stroboscopic_populations_exact(Ncells, phi, A, w, initial_site, ncycles, zeta=zeta, J=J)
```

```python
save_data(DATA / 'FigA04_dynamics_U_site.npz',
          parameters={'J': J, 'phi': phi, 'Ncells': Ncells, 'w': w, 'A': A, 'zeta': zeta,
                      'ncycles': ncycles, 'initial_site': initial_site},
          tvals=tvals, populations=populations)
```
