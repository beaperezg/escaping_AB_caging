# Fig. 2 Quasienergy spectrum vs driving frequency

Quasienergies from the exact Floquet operator of the square wave (`square_wave_floquet_operator`).

Runtime: ~1 min.

Output: `Data/Fig02_quasienergies.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from tqdm.auto import tqdm

from diamond_chain import driven_bulk_quasienergies, driven_finite_quasienergies
from data_io import save_data
```

```python
J = 1.0
phi = np.pi
Ncells = 14
Atilde = 1.2                                  # A / w is kept fixed
wvals = np.linspace(2, 10, 450)
kvals = np.linspace(-np.pi, np.pi, 151)
```

```python
results = {}

for zeta, label in [(-1, 'antisymmetric'), (+1, 'symmetric')]:
    qk_max, qk_min, qe_finite = [], [], []

    for w in tqdm(wvals, desc=label):
        A = Atilde * w
        qe_bulk = driven_bulk_quasienergies(kvals, phi, A, w, zeta=zeta, J=J)
        qk_max.append(qe_bulk.max(axis=0))
        qk_min.append(qe_bulk.min(axis=0))
        qe_finite.append(driven_finite_quasienergies(Ncells, phi, A, w, zeta=zeta, J=J, ending='closed'))

    results[f'qk_max_{label}'] = np.array(qk_max)
    results[f'qk_min_{label}'] = np.array(qk_min)
    results[f'qe_finite_{label}'] = np.array(qe_finite)
```

```python
save_data(DATA / 'Fig02_quasienergies.npz',
          parameters={'J': J, 'phi': phi, 'Ncells': Ncells, 'Atilde': Atilde, 'nk': len(kvals)},
          wvals=wvals, **results)
```
