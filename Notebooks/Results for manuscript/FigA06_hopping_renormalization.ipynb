# Fig. A6 — Hopping renormalization in the effective model

Effective first- and second-neighbor hoppings $J_1(\tilde A)$ and $J_2(\tilde A)$, Eqs. (j1)–(j2) of the manuscript,
for $J = 1$ and $\omega = 8$ (cutoffs $n_\mathrm{max} = m_\mathrm{max} = 99$).
$\Gamma$ is the full (positive and negative harmonics) sum of Eq. (gamma_Def); at $\tilde A = 1.2$, $\omega = 8$:
$|J_1| = 0.469$, $|J_2| = 0.0215$, $|J_2/J_1| = 0.046$.

Output: `Data/FigA06_hoppings.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from effective_model import J1, J2
from data_io import save_data
```

```python
J = 1.0
w = 8
n_max, m_max = 99, 99
Atilde_vals = np.linspace(0.001, 10.001, 400)

J1_vals = np.array([J1(At, w, J, n_max, m_max) for At in Atilde_vals])
J2_vals = np.array([J2(At, w, J, n_max, m_max) for At in Atilde_vals])
```

```python
save_data(DATA / 'FigA06_hoppings.npz', parameters={'J': J, 'w': w, 'n_max': n_max, 'm_max': m_max},
          Atilde_vals=Atilde_vals, J1=J1_vals, J2=J2_vals)
```
