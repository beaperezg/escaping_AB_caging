# Fig. A3 — Convergence properties of $\Gamma$

**a)** $\Gamma$ as a function of the cutoff $n_\mathrm{max}$ for several $m_\mathrm{max}$, at $\tilde A = 1.2$ and $\tilde A = 2.3$.
**b)** Converged $\Gamma(\tilde A)$. **c)** Real and imaginary part of $\Gamma \sin(\pi\tilde A)[1 + e^{i\pi\tilde A}]$ (computed in the plotting notebook).

> **Correction with respect to the original code.** $\Gamma$ is now computed exactly as written in Eq. (gamma_Def) of the Appendix:
> a nested sum over all odd $m$ and all $n \neq 0, m$, **positive and negative**, truncated symmetrically
> ($|n| \le n_\mathrm{max}$, $|m| \le m_\mathrm{max}$, both inclusive). The original code only summed $n, m \geq 1$ (and its
> $-3/\Pi_m^2$ sum stopped at $m_\mathrm{max} - 2$), giving $\Gamma(1.2) = -18.56$ instead of the correct $-49.01$. Only the full
> sum reproduces the bandwidth of the exact Floquet spectrum.
> All panels, and Fig. A6, change accordingly.

Output: `Data/FigA03_convergence_Gamma.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from effective_model import compute_Gamma
from data_io import save_data
```

```python
# a) Gamma vs n_max for several m_max
n_max_vals = np.arange(0, 19)
m_max_vals = np.array([1, 3, 5, 7])
Atilde_conv = np.array([1.2, 2.3])

Gamma_vs_cutoff = np.array([[[compute_Gamma(At, n_max, m_max) for n_max in n_max_vals]
                             for m_max in m_max_vals] for At in Atilde_conv])     # (Atilde, m_max, n_max)

# b) converged Gamma vs Atilde
n_max, m_max = 39, 39
Atildevals = np.linspace(0.001, 2.999, 201)
Gamma = np.array([compute_Gamma(At, n_max, m_max) for At in Atildevals])
```

```python
save_data(DATA / 'FigA03_convergence_Gamma.npz', parameters={'n_max': n_max, 'm_max': m_max},
          n_max_vals=n_max_vals, m_max_vals=m_max_vals, Atilde_conv=Atilde_conv, Gamma_vs_cutoff=Gamma_vs_cutoff,
          Atildevals=Atildevals, Gamma=Gamma)
```
