# Fig. A7 — Diamond chain with second-neighbor hoppings vs SSH chain

Spectra as a function of $J_2/J_1$ of
**a), c)** a closed diamond chain with real first- ($J_1 = 1$) and second-neighbor ($J_2$) hoppings, with $N = 20$ and $N = 21$ cells;
**b), d)** an SSH chain with hoppings $2J_1$, $2J_2$ ($2J_1 = 2$), with $2N$ and $2N+1$ sites ($N = 20$).
The shaded region is the effective bulk band, Eq. (bulk).

Output: `Data/FigA07_diamond_vs_SSH.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from effective_model import diamond_chain_second_neighbors, ssh_chain, eigenvalues_vs_parameter, effective_band_edges
from data_io import save_data
```

```python
# Diamond chain, J1 = 1
J1 = 1.0
J2_diamond = np.linspace(0, 1.5, 100)
E_diamond_even = eigenvalues_vs_parameter(diamond_chain_second_neighbors, 'J2', J2_diamond, Ncells=20, J1=J1)
E_diamond_odd = eigenvalues_vs_parameter(diamond_chain_second_neighbors, 'J2', J2_diamond, Ncells=21, J1=J1)
band_max, band_min = effective_band_edges(J1, J2_diamond)

# SSH chain with hoppings t1 = 2 J1 = 2 and t2 = 2 J2
t1 = 2.0
t2_ssh = np.linspace(0, 3, 201)
E_ssh_even = eigenvalues_vs_parameter(ssh_chain, 't2', t2_ssh, Ncells=20, t1=t1, extra_site=False)
E_ssh_odd = eigenvalues_vs_parameter(ssh_chain, 't2', t2_ssh, Ncells=20, t1=t1, extra_site=True)
```

```python
save_data(DATA / 'FigA07_diamond_vs_SSH.npz',
          parameters={'J1': J1, 'Ncells_even': 20, 'Ncells_odd': 21, 't1_ssh': t1, 'Ncells_ssh': 20},
          J2_diamond=J2_diamond, E_diamond_even=E_diamond_even, E_diamond_odd=E_diamond_odd,
          band_max=band_max, band_min=band_min,
          t2_ssh=t2_ssh, E_ssh_even=E_ssh_even, E_ssh_odd=E_ssh_odd)
```
