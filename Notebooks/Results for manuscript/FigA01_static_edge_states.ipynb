# Fig. A1 — Edge states of the static chain: closed vs open end

Edge-state profiles $|\psi_i|^2$ for **a)** a closed and **b)** an open chain with $N = 6$ cells, $J = 1$, $\phi = \pi$.

The closed chain has two degenerate edge states per gap ($E = \pm\sqrt{2}J$). Inside a degenerate pair `eigh` returns an
arbitrary basis, so the pair is rotated into the states localized on the left and right ends
(the original code relied on the basis returned by `eigh`).

Output: `Data/FigA01_static_edge_states.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from diamond_chain import finite_chain, localize_degenerate_pair
from data_io import save_data
```

```python
J = 1.0
phi = np.pi
Ncells = 6

# Closed chain: edge states at indices (N-1, N) for E = -sqrt(2) J and (2N, 2N+1) for E = +sqrt(2) J
E_closed, vecs_closed = np.linalg.eigh(finite_chain(Ncells, phi, J, ending='closed'))
lower = localize_degenerate_pair(vecs_closed[:, [Ncells - 1, Ncells]])
upper = localize_degenerate_pair(vecs_closed[:, [2 * Ncells, 2 * Ncells + 1]])
print('closed-chain edge energies:', E_closed[[Ncells - 1, Ncells, 2 * Ncells, 2 * Ncells + 1]])

# Open chain: one edge state per gap, indices N-1 and 2N
E_open, vecs_open = np.linalg.eigh(finite_chain(Ncells, phi, J, ending='open'))
print('open-chain edge energies:', E_open[[Ncells - 1, 2 * Ncells]])
```

```python
save_data(DATA / 'FigA01_static_edge_states.npz', parameters={'J': J, 'phi': phi, 'Ncells': Ncells},
          closed_lower_left=lower[:, 0], closed_lower_right=lower[:, 1],
          closed_upper_left=upper[:, 0], closed_upper_right=upper[:, 1],
          open_lower=vecs_open[:, Ncells - 1], open_upper=vecs_open[:, 2 * Ncells])
```
