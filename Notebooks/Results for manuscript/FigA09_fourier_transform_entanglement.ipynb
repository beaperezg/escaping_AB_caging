# Fig. A9 — Fourier transform of the entanglement entropy vs $V$

Power spectrum of $S^{(A)}(mT)$ over 40000 driving cycles as a function of the interaction $V \in [0, 0.2]$,
with the parameters of Fig. 7 ($N = 2$, $\omega = 8$, $A = 9.6$, antisymmetric driving).

> **Partition.** The code (here and in the original) uses $A = \{C_1, U_1, D_1\}$ and $B = \{C_2, U_2, D_2, C_3\}$.
> This is the partition defined in the manuscript (Sec. V).

> The data in `Data/FigA09_FT_entanglement.npz` was computed with this notebook (about 20 minutes with 11 cores).
> It reproduces the original output of `Python/Two particles/2try_two_particle.ipynb` (`FT_colormap.npz`): same peak
> frequencies, total power per $V$ within 1.3%.

Output: `Data/FigA09_FT_entanglement.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from two_particles import generate_psi0, fourier_transform_of_entanglement
from parallel_utils import parameter_run
from data_io import save_data
```

```python
Ncells = 2                                    # 7 sites: C1 U1 D1 C2 U2 D2 C3
Nup, Ndown = 2, 0                             # two same-spin fermions
J = 1.0
phi = np.pi
w = 8
zeta = -1
A_spec = [('C', 1), ('U', 1), ('D', 1)]       # subsystem A; B is the complement {C2, U2, D2, C3}
psi0_binary = '1000001|0000000'               # one fermion on C1 and one on C3
psi0 = generate_psi0(psi0_binary, Ncells, Nup, Ndown)

A = 9.6
ncycles = 40000
Vvals = np.linspace(0, 0.2, 300)
```

```python
freqs, FT_S_A, FT_E_P = parameter_run(fourier_transform_of_entanglement, 'V', Vvals,
                                      Ncells=Ncells, Nup=Nup, Ndown=Ndown, A=A, w=w, psi0=psi0,
                                      A_spec=A_spec, ncycles=ncycles, zeta=zeta, J=J, phi=phi)
```

```python
save_data(DATA / 'FigA09_FT_entanglement.npz',
          parameters={'Ncells': Ncells, 'Nup': Nup, 'Ndown': Ndown, 'J': J, 'phi': phi, 'w': w, 'A': A,
                      'zeta': zeta, 'A_spec': A_spec, 'psi0_binary': psi0_binary, 'ncycles': ncycles},
          Vvals=Vvals, freqs=freqs[0], FT_S_A=FT_S_A, FT_E_P=FT_E_P)
```
