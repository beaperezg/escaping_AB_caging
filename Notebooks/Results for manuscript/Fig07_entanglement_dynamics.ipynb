# Fig. 7 — Two-particle entanglement dynamics

Two same-spin fermions on a closed chain with $N = 2$ cells (7 sites), initial state $C_1^\dagger C_3^\dagger |0\rangle$,
antisymmetric driving with $\omega = 8$, $\tilde A = 1.2$ ($A = 9.6$), $J = 1$, $\phi = \pi$,
nearest-neighbor interaction **a, b)** $V = 0$ and **c, d)** $V = 0.2$.

Computed at stroboscopic times (400 cycles) with the exact Floquet operator of the square wave
(`stroboscopic_states_exact`): $S^{(A)}$, $p_{1,1}$, $E_P(1,1) = p_{1,1} S^{(A)}_{1,1}$ and the site densities.

> **Partition.** The code (here and in the original) uses $A = \{C_1, U_1, D_1\}$ and $B = \{C_2, U_2, D_2, C_3\}$.
> This is the partition defined in the manuscript (Sec. V).

Output: `Data/Fig07_entanglement_V0.npz`, `Data/Fig07_entanglement_V0.2.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from two_particles import (generate_psi0, stroboscopic_states_exact,
                           entanglement_observables, site_densities)
from data_io import save_data
```

```python
Ncells = 2                                    # 7 sites: C1 U1 D1 C2 U2 D2 C3
Nup, Ndown = 2, 0                             # two same-spin fermions
J = 1.0
phi = np.pi
w = 8
zeta = -1
A_spec = [('C', 1), ('U', 1), ('D', 1)]       # subsystem A; B is the complement {C2, U2, D2, C3}
psi0_binary = '1000001|0000000'               # one fermion on C1 and one on C3
psi0 = generate_psi0(psi0_binary, Ncells, Nup, Ndown)

A = 9.6
ncycles = 400
```

```python
for V in (0.0, 0.2):
    tvals, states = stroboscopic_states_exact(Ncells, Nup, Ndown, V, A, w, psi0, ncycles, zeta=zeta, J=J, phi=phi)
    observables = entanglement_observables(states, Ncells, Nup, Ndown, A_spec)
    densities = site_densities(states, Ncells, Nup, Ndown)

    save_data(DATA / f'Fig07_entanglement_V{V:g}.npz',
              parameters={'Ncells': Ncells, 'Nup': Nup, 'Ndown': Ndown, 'J': J, 'phi': phi, 'w': w, 'A': A,
                          'zeta': zeta, 'V': V, 'A_spec': A_spec, 'psi0_binary': psi0_binary, 'ncycles': ncycles},
              tvals=tvals, densities=densities, **observables)
```
