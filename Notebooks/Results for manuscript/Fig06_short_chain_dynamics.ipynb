# Fig. 6 — Dynamics in short chains

Stroboscopic occupations for closed chains with **a)** $N = 2$ (7 sites) and **b)** $N = 3$ (10 sites), particle initially on $C_1$,
antisymmetric driving, $\omega = 8$, $\tilde A = 1.2$ ($A = 9.6$), $J = 1$, $\phi = \pi$, 500 cycles.
Computed with the exact Floquet operator of the square wave (`stroboscopic_populations_exact`).

Output: `Data/Fig06_short_chain_dynamics.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from diamond_chain import site_index, stroboscopic_populations_exact
from data_io import save_data
```

```python
J = 1.0
phi = np.pi
w = 8
A = 9.6
zeta = -1
ncycles = 500

populations = {}
for Ncells in (2, 3):
    # Exact Floquet operator of the square wave (two matrix exponentials per period)
    tvals, populations[f'populations_N{Ncells}'] = stroboscopic_populations_exact(Ncells, phi, A, w, site_index('C', 1),
                                                                                  ncycles, zeta=zeta, J=J)
```

```python
save_data(DATA / 'Fig06_short_chain_dynamics.npz',
          parameters={'J': J, 'phi': phi, 'w': w, 'A': A, 'zeta': zeta, 'ncycles': ncycles, 'initial_site': 'C_1'},
          tvals=tvals, **populations)
```
