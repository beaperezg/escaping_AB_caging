# Fig. A10 — Maximum entanglement as a function of $A$ and $V$

Maximum over the first 200 driving cycles of **a)** $S^{(A)}$ and **b)** $E_P(1,1)$, as a function of the driving amplitude
$A \in [0, 20]$ and the interaction $V \in [0, 0.5]$, with the other parameters as in Fig. 7.

> **Inconsistency found in the original code.** The original function `maximum_entanglement` returned the maximum of
> $S^{(A)}_{1,1}$ (the entropy inside the $(1,1)$ sector) instead of $E_P(1,1) = p_{1,1} S^{(A)}_{1,1}$, so the published
> panel b) showed $\max S^{(A)}_{1,1}$ (checked: e.g. at $V = 0.201$, $A = 9.65$ the stored value is 0.694 $= \max S_{1,1}$,
> while $\max E_P = 0.680$). This notebook stores the three quantities `max_S_A`, `max_E_P` and `max_S_11`.

> **Partition.** The code (here and in the original) uses $A = \{C_1, U_1, D_1\}$ and $B = \{C_2, U_2, D_2, C_3\}$.
> This is the partition defined in the manuscript (Sec. V).

> The data in `Data/FigA10_max_entanglement.npz` was computed with this notebook (about 15 minutes with 11 cores).
> The recomputed `max_S_A` and `max_S_11` agree with the original output of `Python/Two particles/2try_two_particle.ipynb`
> to within 3·10⁻³.

Output: `Data/FigA10_max_entanglement.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from two_particles import generate_psi0, max_entanglement
from parallel_utils import parameter_run
from data_io import save_data
```

```python
Ncells = 2                                    # 7 sites: C1 U1 D1 C2 U2 D2 C3
Nup, Ndown = 2, 0                             # two same-spin fermions
J = 1.0
phi = np.pi
w = 8
zeta = -1
A_spec = [('C', 1), ('U', 1), ('D', 1)]       # subsystem A; B is the complement {C2, U2, D2, C3}
psi0_binary = '1000001|0000000'               # one fermion on C1 and one on C3
psi0 = generate_psi0(psi0_binary, Ncells, Nup, Ndown)

ncycles = 200
Vvals = np.linspace(0, 0.5, 200)
Avals = np.linspace(0, 20, 200)
```

```python
max_S_A, max_E_P, max_S_11 = parameter_run(max_entanglement, [['V'], ['A']], [[Vvals], [Avals]],
                                           Ncells=Ncells, Nup=Nup, Ndown=Ndown, w=w, psi0=psi0,
                                           A_spec=A_spec, ncycles=ncycles, zeta=zeta, J=J, phi=phi)
```

```python
save_data(DATA / 'FigA10_max_entanglement.npz',
          parameters={'Ncells': Ncells, 'Nup': Nup, 'Ndown': Ndown, 'J': J, 'phi': phi, 'w': w,
                      'zeta': zeta, 'A_spec': A_spec, 'psi0_binary': psi0_binary, 'ncycles': ncycles},
          Vvals=Vvals, Avals=Avals,                       # arrays below have shape (len(Vvals), len(Avals))
          max_S_A=max_S_A[..., 0], max_E_P=max_E_P[..., 0], max_S_11=max_S_11[..., 0])
```
