# Fig. A5 — Dynamics from the central $C$ site for several driving frequencies

Stroboscopic occupations of a closed chain with $N = 100$ cells, particle initially on $C_{51}$, antisymmetric driving at fixed
$\tilde A = A/\omega = 1.2$, for $\omega = 8, 6, 4, 3, 2$ ($J = 1$, $\phi = \pi$, 400 cycles). The figure shows $\omega = 8, 6, 4$.
Computed with the exact Floquet operator of the square wave (`stroboscopic_populations_exact`).

Output: `Data/FigA05_dynamics_C_site_vs_w.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from diamond_chain import site_index, stroboscopic_populations_exact
from data_io import save_data
```

```python
J = 1.0
phi = np.pi
Ncells = 100
w_values = np.array([8.0, 6.0, 4.0, 3.0, 2.0])
A_over_w = 1.2
zeta = -1
ncycles = 400

initial_site = site_index('C', 51)

all_tvals = []
all_populations = []
for w in w_values:
    # Exact Floquet operator of the square wave (two matrix exponentials per period)
    tvals, populations = stroboscopic_populations_exact(Ncells, phi, A_over_w * w, w, initial_site, ncycles, zeta=zeta, J=J)
    all_tvals.append(tvals)
    all_populations.append(populations)
```

```python
save_data(DATA / 'FigA05_dynamics_C_site_vs_w.npz',
          parameters={'J': J, 'phi': phi, 'Ncells': Ncells, 'w_values': w_values, 'A_over_w': A_over_w, 'zeta': zeta,
                      'ncycles': ncycles, 'initial_site': initial_site},
          tvals=np.array(all_tvals), populations=np.array(all_populations))
```
