# Fig. 5 — Edge states of the driven chain

Quasienergies and Floquet modes (at $t = 0$) of a closed chain with $N = 14$ cells, antisymmetric driving,
$\omega = 8$, $\tilde A = 1.2$ ($A = 9.6$), $J = 1$, $\phi = \pi$.

The edge states are the quasienergy indices $N-1,\ N$ (lower gap) and $2N,\ 2N+1$ (upper gap).
Quasienergies and modes from the exact Floquet operator of the square wave (`square_wave_floquet_operator`).
Each pair is split by only $\sim 10^{-7}$; the exact operator resolves this splitting (the former QuTiP time
integration, with errors $\sim 10^{-5}$, did not), so the modes within a pair are the bonding/antibonding
combinations of the left and right edge states.

Output: `Data/Fig05_driven_edge_states.npz`

```python
from pathlib import Path

import numpy as np

ROOT = Path.cwd().parents[1]                 # .../paper_reproduction
DATA = ROOT / 'Data'

from diamond_chain import driven_finite_quasienergies
from data_io import save_data
```

```python
J = 1.0
phi = np.pi
Ncells = 14
w = 8
A = 9.6
zeta = -1

quasienergies, modes = driven_finite_quasienergies(Ncells, phi, A, w, zeta=zeta, J=J, ending='closed', return_modes=True)

edge = [Ncells - 1, Ncells, 2 * Ncells, 2 * Ncells + 1]
print('edge-state quasienergies:', quasienergies[edge])
```

```python
save_data(DATA / 'Fig05_driven_edge_states.npz',
          parameters={'J': J, 'phi': phi, 'Ncells': Ncells, 'w': w, 'A': A, 'zeta': zeta},
          quasienergies=quasienergies, modes=modes)
```
